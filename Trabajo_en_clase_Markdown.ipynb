#**INFORME VALOR PRESENTE NETO**

   **Valor presente neto** o ***VPN*** es una herramienta que permite determinar el valor actual de los flujos de caja futuros estimados en un proyecto, descontando una tasa que representa costos financieros, impuestos, entre otros conceptos, permitiendo traer al tiempo presente el valor neto de cada flujo proyectado para poder evaluar su rentabilidad al compararse con la inversión inicial.

  El valor presente neto **_se calcula_** como la diferencia entre el valor actual de los flujos de efectivo futuros esperados de una inversión y el costo inicial de la inversión. Es decir, ***___el VPN permite conocer si el valor actual de los ingresos generados por un proyecto supera el costo actual de su ejecución___***.




Ademas

- **Inversión inicial:** Es el monto que debe ser invertido en el proyecto para
generar los ingresos futuros.
- **Tasa de descuento:** Es el porcentaje de impacto que tendrán los costos financieros, impuestos y costo de oportunidad, entre otros, en los ingresos futuros.
- **Flujo de caja o ingresos:** Es el monto de los ingresos estimados que generará cada periodo proyectado.
- **Periodo o duración proyectada:** Es el número de periodos, generalmente años, que se han proyectado para la generación del ingreso a evaluar. Por ejemplo, 5 años o 10 años.

##**___FÓRMULA VALOR PRESENTE NETO (VPN)___**

La fórmula para calcular el valor presente es la siguiente:

\begin{equation}
    \text{VPN} = \sum_{t=1}^{n} \frac{FC_t}{(1+r)^t} - I_0
\end{equation}
