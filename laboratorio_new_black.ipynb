### 🧩 Tu turno 2.1 — Rendimientos diarios

El **rendimiento logarítmico** mide cuánto cambió el precio de un día al siguiente:

```python
rendimiento_t = ln(Precio_hoy / Precio_ayer)
```

En código: `np.log(precios / precios.shift(1))`

Calcula los rendimientos de Bitcoin y grafícalos como barras (verde = sube, rojo = baja).


````python
# ── Celda 2.1 — Rendimientos diarios ───────────────────────────────────────
precios_btc = datos["Close"]["BTC-USD"].dropna()

# Calcula los rendimientos logarítmicos
rend_btc = np.log(precios_btc / precios_btc.shift(1))

# Gráfica de barras
colores_barras = ["#27ae60" if r >= 0 else "#e74c3c" for r in rend_btc.values]

fig = go.Figure()
fig.add_trace(go.Bar(
    x      = rend_btc.index,
    y      = rend_btc.values,
    marker_color = colores_barras,
    name   = "Rendimiento diario"
))
fig.add_hline(y=0, line_color="black", line_width=1)
fig.update_layout(
    title       = "Rendimientos diarios — Bitcoin",
    xaxis_title = "Fecha",
    yaxis_title = "Log-retorno",
    height      = 350,
    showlegend  = False
)
fig.show()

print(f"Días analizados:   {len(rend_btc)}")
print(f"Días positivos:    {(rend_btc > 0).sum()}  ({(rend_btc > 0).mean()*100:.1f}%)")
print(f"Días negativos:    {(rend_btc < 0).sum()}  ({(rend_btc < 0).mean()*100:.1f}%)")
````

````python
#@title ✅  Solución 2.1 — Rendimientos diarios  —  haz clic en ► para ver

precios_btc = datos["Close"]["BTC-USD"].dropna()
rend_btc    = np.log(precios_btc / precios_btc.shift(1)).dropna()

colores_barras = ["#27ae60" if r >= 0 else "#e74c3c" for r in rend_btc.values]

fig = go.Figure()
fig.add_trace(go.Bar(x=rend_btc.index, y=rend_btc.values,
    marker_color=colores_barras, name="Rendimiento diario",
    hovertemplate="<b>%{x}</b><br>Retorno: %{y:.4f}<extra></extra>"))
fig.add_hline(y=0, line_color="black", line_width=1)
fig.update_layout(title="Rendimientos diarios — Bitcoin",
    xaxis_title="Fecha", yaxis_title="Log-retorno",
    height=350, showlegend=False,
    paper_bgcolor="rgba(0,0,0,0)", plot_bgcolor="#f8f9fa")
fig.show()

dias_pos = (rend_btc > 0).sum(); dias_neg = (rend_btc < 0).sum()
print(f"📊  Estadísticas de rendimientos:")
print(f"   Días positivos: {dias_pos}  ({dias_pos/len(rend_btc)*100:.1f}%)")
print(f"   Días negativos: {dias_neg}  ({dias_neg/len(rend_btc)*100:.1f}%)")
print(f"   Media diaria:   {rend_btc.mean()*100:.4f}%")
print(f"   Desv. estándar: {rend_btc.std()*100:.4f}%")
````

### 🧩 Tu turno 2.2 — Rendimiento acumulado

Si un inversionista hubiera puesto **$10,000** el primer día, ¿cuánto tendría hoy?

```python
valor = CAPITAL_INICIAL * np.exp(rendimientos.cumsum())
```

Calcula esto para las **5 criptos** y grafícalas juntas. Esta visualización es la primera
que los inversionistas van a pedir ver en el stand.


````python
fig = go.Figure()

for cripto, nombre, color in zip(CRIPTOS, NOMBRES, COLORES):
    precios = datos["Close"][cripto].dropna()
    rend    = np.log(precios / precios.shift(1)).dropna()

    # Calcula el valor acumulado de la inversión
    valor_acum = CAPITAL_INICIAL * np.exp(rend.cumsum())

    retorno_pct = (valor_acum.iloc[-1] / CAPITAL_INICIAL - 1) * 100

    fig.add_trace(go.Scatter(
        x    = valor_acum.index,
        y    = valor_acum.values,
        name = f"{nombre}  (${valor_acum.iloc[-1]:,.0f}  |  {retorno_pct:+.0f}%)",
        line = dict(color=color, width=2)
    ))

fig.add_hline(y=CAPITAL_INICIAL, line_dash="dot", line_color="gray",
              annotation_text=f"Capital inicial ${CAPITAL_INICIAL:,}")
fig.update_layout(
    title       = f"¿Cuánto valdría ${CAPITAL_INICIAL:,} invertidos en cada cripto?",
    xaxis_title = "Fecha",
    yaxis_title = "Valor del portafolio (USD)",
    height      = 450,
    hovermode   = "x unified",
    legend      = dict(orientation="h", yanchor="bottom", y=1.02)
)
fig.show()
````

````python
fig = go.Figure()
resultados_acum = {}

for cripto, nombre, color in zip(CRIPTOS, NOMBRES, COLORES):
    precios    = datos["Close"][cripto].dropna()
    rend       = np.log(precios / precios.shift(1)).dropna()
    valor_acum = CAPITAL_INICIAL * np.exp(rend.cumsum())
    resultados_acum[nombre] = valor_acum.iloc[-1]
    retorno_pct = (valor_acum.iloc[-1] / CAPITAL_INICIAL - 1) * 100

    fig.add_trace(go.Scatter(
        x=valor_acum.index, y=valor_acum.values,
        name=f"{nombre}  (${valor_acum.iloc[-1]:,.0f}  |  {retorno_pct:+.0f}%)",
        line=dict(color=color, width=2.2),
        hovertemplate=f"<b>{nombre}</b><br>$%{{y:,.0f}}<extra></extra>"))

fig.add_hline(y=CAPITAL_INICIAL, line_dash="dot", line_color="gray",
              annotation_text=f"Capital inicial ${CAPITAL_INICIAL:,}")
fig.update_layout(
    title=f"¿Cuánto valdría ${CAPITAL_INICIAL:,} invertidos en cada cripto?<br>"
          f"<sup>{FECHA_INICIO} → {FECHA_FIN}</sup>",
    xaxis_title="Fecha", yaxis_title="Valor del portafolio (USD)",
    height=450, hovermode="x unified",
    legend=dict(orientation="h", yanchor="bottom", y=1.02),
    paper_bgcolor="rgba(0,0,0,0)", plot_bgcolor="#f8f9fa")
fig.show()

print("\n📊  Ranking de rendimiento total:")
for i,(n,v) in enumerate(sorted(resultados_acum.items(),key=lambda x:-x[1]),1):
    print(f"  {i}. {n:<12} ${v:>10,.0f}  ({(v/CAPITAL_INICIAL-1)*100:+.0f}%)")
````

### 🧩 Tu turno 2.3 — Tabla de métricas de riesgo

Construye una tabla con las 5 métricas obligatorias del concurso para las 5 criptos.

| Métrica | Fórmula |

| Maximum Drawdown | caída máxima desde el pico |
| % días negativos | `(rendimientos < 0).mean() × 100` |


````python
filas = []

for cripto, nombre in zip(CRIPTOS, NOMBRES):
    precios   = datos["Close"][cripto].dropna()
    rend_log  = np.log(precios / precios.shift(1)).dropna()
    rend_simp = precios.pct_change().dropna()

    retorno_total  = (precios.iloc[-1] / precios.iloc[0] - 1) * 100   # %
    volatilidad    = rend_log.std() * np.sqrt(252) # desv. estándar anualizada
    riqueza        = (1 + rend_simp).cumprod()
    drawdown       = (riqueza - riqueza.cummax()) / riqueza.cummax()
    max_drawdown   = drawdown.min()   # peor caída desde el pico
    pct_negativos  = (rend_log < 0).mean() * 100  # % días con rend. negativo

    filas.append({
        "Criptomoneda":           nombre,
        "Retorno total (%)":      round(retorno_total, 1),
        "Volatilidad anual (%)":  round(volatilidad * 100, 1),
        "Max Drawdown (%)":       round(max_drawdown * 100, 1),
        "% Días negativos":       round(pct_negativos, 1),
    })

tabla_metricas = pd.DataFrame(filas).set_index("Criptomoneda")

# Muestra con color (verde = mejor, rojo = peor)
tabla_metricas.style\
    .background_gradient(subset=["Retorno total (%)"], cmap="RdYlGn")\
    .background_gradient(subset=["Volatilidad anual (%)"], cmap="YlOrRd")\
    .background_gradient(subset=["Max Drawdown (%)"], cmap="RdYlGn")\
    .background_gradient(subset=["% Días negativos"], cmap="YlOrRd")\
    .format("{:.1f}")
````

### ⚠️ La trampa del lookahead bias

El error más común en backtesting es usar la señal de **hoy** para comprar **hoy**.
En la realidad, el precio de cierre de hoy solo lo sabes al final del día.
La señal de hoy se aplica **mañana**.

```python
# ❌ INCORRECTO — usas información que no tenías
rend_estrategia = rendimientos * señal

# ✅ CORRECTO — la señal de hoy aplica mañana
rend_estrategia = rendimientos * señal.shift(1)
```
