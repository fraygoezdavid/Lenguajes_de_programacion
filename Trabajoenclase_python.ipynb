# Simulación básica de inversion
## Vamos a modelar un proyecto de inversion


# Declarar las variables financieras

```python
inversion_inicial = 100000000 # Esta variable es de tipo entero
```

```python
tasa_descuento = 0.10 # Esta variable es de tipo <class 'float'>
```

```python
# Definiendo flujos
flujo_1 = 1500000
flujo_2 = 2000000
flujo_3 = 1000000
flujo_4 = 2500000
```

```python
#verificando el tipo y valor de la variable
print(type(inversion_inicial))
print(type(tasa_descuento))
print (type(flujo_1))
```

#**OPERACIÓN BÁSICA FINANCIERA**

```python
total_flujos = flujo_1 + flujo_2 + flujo_3 + flujo_4
```

```python
print("Total de flujos recibidos:", total_flujos)
```

```python
print(type(total_flujos))
```

#Compraciones

```python
proyecto_rentable = total_flujos > inversion_inicial
```

```python
print("El proyecto recupera la inversion:", proyecto_rentable)
```

```python
print (type(proyecto_rentable))
```

```python
nombre_proyecto = "Inversión Propia"

rentabilidad_esperada = 2.5
numero_inversinistas = 4
proyecto_aprobado = total_flujos > inversion_inicial
```

```python
print(type(nombre_proyecto))
print(type(rentabilidad_esperada))
print(type(numero_inversinistas))
print(type(proyecto_aprobado))
```

```python
print("¿El proyecto es rentable para los inversionistas?:", proyecto_aprobado)
```
