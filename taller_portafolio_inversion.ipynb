## Parte 1 — Tuplas (Información fija)
Las tuplas son útiles cuando los datos **no cambian**.

```python
empresas = ("Apple", "Microsoft", "Amazon", "Tesla", "Google")
print(empresas)
```

### Ejercicio
Imprime:

- La tercera empresa

```python
# Escribe tu código aquí
print("empresa numero 1:", empresas[0])
print("empresa numero 2:",empresas[2])
print("empresa numero 3:",empresas[-1])
```

## Parte 2 — Lista (Precio de las acciones)

```python
precios_acciones = [185, 320, 145, 250, 275]
print(precios_acciones)
```

```python
#precio de las acciones
print("Precio de Apple:", precios_acciones[0])
print("Precio de Microsoft:", precios_acciones[1])
print("Precio de Tesla:", precios_acciones[-1])
print("Cantidad de empresas en el portafolio:", len(precios_acciones))
```

## Parte 3 — Diccionario (Cantidad de acciones)

```python
acciones_portafolio = {
    "Apple": 10,
    "Microsoft": 5,
    "Amazon": 8,
    "Tesla": 4,
    "Google": 6
}
print(acciones_portafolio)
```

### Ejercicio
Imprime:
- Acciones de Apple
- Acciones de Google

```python
# Escribe tu código aquí
print("acciones de Apple:", acciones_portafolio["Apple"])
print("acciones de Google:", acciones_portafolio["Google"])
```

## Parte 4 — Valor invertido

```python
valor_apple = precios_acciones[0] * acciones_portafolio["Apple"]
print("Valor invertido en apple:", valor_apple)
```

### Ejercicio
Calcula el valor invertido en Microsoft, Amazon, Tesla y Google.

```python
# Escribe tu código aquí
print("valor invertido en Microsoft:", precios_acciones[1] * acciones_portafolio["Microsoft"])
print("valor invertido en Amazon:", precios_acciones[2] * acciones_portafolio["Amazon"])
print("valor invertido en Tesla:", precios_acciones[3] * acciones_portafolio["Tesla"])
print("valor invertido en Google:", precios_acciones[4] * acciones_portafolio["Google"])
```

## Parte 5 — Lista del portafolio

```python
valor_portafolio = [3]
valor_portafolio.append(valor_apple)
print("Valor portafolio apple:",valor_portafolio)
valor_portafolio.append(precios_acciones[1] * acciones_portafolio["Microsoft"])
print("Valor portafolio Microsoft:",valor_portafolio)


# Agrega los demás valores calculados
```

```python
print("Valor total del portafolio:", sum(valor_portafolio))
print("Promedio:", sum(valor_portafolio)/len(valor_portafolio))
print("Número de empresas:", len(valor_portafolio))
```
