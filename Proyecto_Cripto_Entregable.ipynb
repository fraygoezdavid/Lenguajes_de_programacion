#Riesgo: volatilidad anualizada
- La volatilidad mide cuánto varía un activo. Anualizarla permite comparar entre activos y horizontes.
- Calcula la volatilidad anualizada (desviación estándar × √252 días bursátiles).

```python
vol_anual = btc['retorno'].std(1) * np.sqrt(252)
```

#Visualización de precios y retornos (Graficos)
- Un buen gráfico vale más que mil tablas — el jurado lo agradecerá.

```python
# Línea de precios
fig = px.line(df, y='Close', title='Bitcoin 2024')
fig.show()

# Histograma de retornos
fig2 = px.histogram(df, x='retorno', nbins=50)
fig2.show()
```

#Matriz de correlación
- Si dos cripto se mueven igual (corr ≈ 1), no diversificas. Busca pares con baja correlación.

```python
# Calcular correlación
returns = pd.DataFrame({
  'BTC': btc['Close'].pct_change(),
  'ETH': eth['Close'].pct_change(),
  'SOL': sol['Close'].pct_change(),
  'ADA': ada['Close'].pct_change(),
})
corr = returns.corr()

import seaborn as sns
sns.heatmap(corr, annot=True, cmap='coolwarm')
```
