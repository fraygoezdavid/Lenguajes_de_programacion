#Maximum Drawdown
“El Maximum Drawdown representa la peor caída histórica desde un pico hasta un mínimo posterior.”

```python
# USAR SOLO LA COLUMNA CLOSE

precio = btc['Close']

# MAXIMO ACUMULADO
max_acumulado = precio.cummax()

# DRAWDOWN
drawdown = (precio / max_acumulado) - 1

# MAXIMUM DRAWDOWN
max_drawdown = drawdown.min()

print("Maximum Drawdown:")
print(max_drawdown)
```

```python
# GRÁFICO DRAWDOWN

plt.figure(figsize=(14,6))

plt.plot(
    btc.index,
    btc['Drawdown']
)

plt.title("Drawdown Bitcoin")
plt.xlabel("Fecha")
plt.ylabel("Drawdown")

plt.show()
```

#Matriz de correlación
- Si dos cripto se mueven igual (corr ≈ 1), no diversificas. Busca pares con baja correlación.

```python
# Calcular correlación
returns = pd.DataFrame({
  'BTC': btc['Close'].pct_change(),
  'ETH': eth['Close'].pct_change(),
  'SOL': sol['Close'].pct_change(),
  'ADA': ada['Close'].pct_change(),
})
corr = returns.corr()

import seaborn as sns
sns.heatmap(corr, annot=True, cmap='coolwarm')
```
