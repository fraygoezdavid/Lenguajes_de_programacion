###_¿Qué es este concurso?_
Cada equipo actúa como un fondo de inversión en criptomonedas. Su misión es construir una aplicación
interactiva que analice datos históricos, proponga una idea de trading y la defienda ante inversionistas
reales el día del evento.
Los asistentes al evento son los inversionistas. Ellos recorren los stands, ven la app en vivo, escuchan
el pitch de cada fondo y deciden dónde poner su capital ficticio. El fondo que mejor explique su propuesta
capta más inversión.
###_Lo que deben construir_
Una aplicación interactiva en R o Python que descargue datos en tiempo real desde Yahoo Finance el
día del evento. No se permiten archivos CSV precargados: la app debe conectarse a la fuente en vivo.
